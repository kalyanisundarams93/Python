**List (Creation, Modifi cation and Access):**

1. List Creation:  
a. Create a list named age_list with five integer elements. For eg., [24, 25, 26, 27, 28]  
b. Create a list named name_list with five string elements.

```python
age_list = [24,25,26,27,28]
name_list = ['Kalyan','Lakshmi','Sundar','Siva','Mani']
print(type(age_list))
print(type(name_list))
```

2. List Operations / Modifications:   
a. Append the string "Yazhini" to name_list.  
b. Insert the element 30 at index 2 in age_list.  
c. Remove the string "Yazhini" from name_list.  
d. Pop the last element from age_list.  
e. Extend the age_list with additional ages [29, 30, 26].  
f. Sort age_list in descending order.  
g. Find Max age, Min age and sum of all ages from age_list.

```python
name_list.append('Yazhini')
print('List with added name: ',name_list)
age_list.insert(2,30)
print('List with inserted age at index 2: ',age_list)
name_list.remove('Yazhini')
print('List with removed name: ',name_list)
age_list.pop(-1)
print('Poping last element from age:', age_list)
age_list.extend([29,30,26])
print('Extending last element from age:', age_list)
age_list.sort(reverse=True)
print('Sorting age list in descending order:', age_list)
print('Max age:', max(age_list) )
print('Min age:', min(age_list))
print('Sum:', sum(age_list))
```

3. Accessing List Elements:  
a. Print the first element of name_list.  
b. Print the last element of name_list.  
c. Print the elements from index 2 to index 4 in name_list.  
d. Print the elements of name_list in reverse order.

```python
print('First element: ', name_list[0])
print('Last element: ', name_list[-1])
print('Elements from idx 2 to 4: ', name_list[2:5])
print('Elements of name_list in reverse order: ', name_list[::-1])
```

**Dictionary (Creation, Modifi cation and Access):**

a. Create a dictionary named student_marks that maps the names of five students to their marks (use scale of from 0 to 100).  
b. Access and print the mark of a specific student, of your choice.  
c. Add a new student "Janani" with a mark of 80 to the student_marks dictionary.  
d. Update the mark of any one older student to 82.  
e. Use the keys(), values(), and items() methods to print all keys, values, and key-value pairs in the student_marks dictionary.

```python
#Create a dictionary named student_marks that maps the names of five students to their marks (use scale of from 0 to 100).
student_marks = {'Arun': 80,
                 'Ravi': 63,
                 'Sam' : 75,
                 'Ram' : 78,
                 'Sundar' : 84
                 }
print(type(student_marks))

#Access and print the mark of a specific student, of your choice.
print(student_marks['Sundar'])

#Add a new student "Janani" with a mark of 80 to the student_marks dictionary.
student_marks["David"] = 95
print(student_marks)

#Update the mark of any one older student to 82.
student_marks["Ram"] = 82
print(student_marks)

#Use the keys(), values(), and items() methods to print all keys, values, and key-value pairs in the student_marks dictionary.
print('Keys: ', student_marks.keys())
print('values: ', student_marks.values())
print('Items: ', student_marks.items())
```

**Sets (Operations):**

a. Create a set called my_set with following values: ['a','e','i','o','u','a','a','i']
Analyse the output and provide explanation for the same.  
b. Attempt to change the value of my_set[4] = 's'. If code throws an error, provide an explanation.  
c. Create two sets: set1 with values: {1, 3, 5, 7, 9} set2 with values: {2, 3, 5, 8, 10}  
d. Compute and print the union and intersection of set1 and set2.

```python
#Create a set called my_set with following values: ['a','e','i','o','u','a','a','i'] Analyse the output and provide explanation for the same.

my_set = {'a','e','i','o','u','a','a','i'}
print(my_set)
```

From the above output, we can see that the sets will eliminate duplicate values  and do not maintain the sequence in which items are entered.

```python
#Attempt to change the value of my_set[4] = 's'. If code throws an error, provide an explanation.
my_set[4] = 's'
print(my_set)
```

We cannot directly change an item that is already inside it.  
Sets are unindexed and mutable but its elements are immutable


```python
#Create two sets: set1 with values: {1, 3, 5, 7, 9} set2 with values: {2, 3, 5, 8, 10}
set1 = {1,3,5,7,9}
set2 = {2,3,5,8,10}
# Compute and print the union and intersection of set1 and set2
print("Union:", set1.union(set2))
print("Intersection:", set1.intersection(set2))
```

Performance Category Program:  
1. Prompt user for Input. Score range should be from 0 to 10 (both inclusive).
2. Find the performance category based on the input score using following criteria:    
      a. Above Average: Score greater than 7  
      b. Average: Score between 4 and 7(both inclusive)  
      c. Below Average: Score lesser than 4
3. Output: Print the Performance category
4. Additional Step: You can give a prompt of your choice to each category.
For eg: If score below average “Need to Improve your performance, consistent practice will lead to better results”.

```python
score = int(input('Enter your score: ' ))
if score > 7:
  print('Above average: Good effort')
elif score >=4 and score <=7 :
  print('Average: Keep practicing')
else:
  print('Below Average: Need to Improve your performance')
```
