PERFORMING DIFFERENT LIST OPERATIONS FROM A LIST OF STUDENT MARKS

marks = [78,85,92,67,85]

```python
# Print all marks
marks = [78,85,92,67,85]
print(marks)
```

```python
# Add a new mark, 90
marks = [78,85,92,67,85]
marks.append(90)
print(marks)
```

```python
# Insert 75 at index 2
marks = [78,85,92,67,85]
marks.insert(2,75)
print(marks)
```

```python
#  Count how many times 85 appears
marks = [78,85,92,67,85]
counts = marks.count(85)
print('Count of 85:' ,counts)
```

```python
# Find the index of 92
marks = [78,85,92,67,85]
idx = marks.index(92)
print('Index of 92: ', idx)
```

```python
# Sort the marks in ascending order
marks = [78,85,92,67,85]
marks.sort()
print(marks)
```

```python
# Display the first three marks using slicing
marks = [78,85,92,67,85]
print(marks[:3])
```

```python
# Display the last two marks using slicing
marks = [78,85,92,67,85]
print(marks[-2:])
```

```python
# Display every second mark using slicing
marks = [78,85,92,67,85]
print(marks[::2])
```

```python
# Display the marks in reverse order using slicing.
marks = [78,85,92,67,85]
print(marks[::-1])
```

```python
#Create a separate copy of the marks list.
marks = [78,85,92,67,85]
new_copy = marks.copy()
print("Original:", marks)
print("New copy:", new_copy)
```

```python
# Remove the lowest mark from the sorted list using pop(0).
marks = [78,85,92,67,85]
marks.sort()
marks.pop(0)
print(marks)
```
