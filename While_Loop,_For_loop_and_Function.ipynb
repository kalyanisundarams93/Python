Number Guessing Game  
Problem Statement: Create a Python program that implements a simple number guessing game using a while loop. The program should make use of control statements such as else, break, and continue

```python
import random
secret_number = random.randint(1,10)
attempts = 3
while attempts > 0:
  guess = int(input('Guess the number between 1 and 10 : '))
  if guess < 1 or guess > 10:
    print('Invalid input! Enter number between 1 and 10')
    continue
  elif guess > secret_number:
    print('Your guess is higher than secret number')
    attempts -=1
  elif guess < secret_number:
    print('Your guess is lower than secret number')
    attempts -=1
  else:
    print('Congrats! You gussed the correct number')
    break
  print('attempts left: ', attempts)
else:
  print('Better luck next time')
```

Multiplication Table Generator  
Problem Statement: Create a Python program that generates and prints a multiplication table (from 1 to 10) for a given number using a for loop and the range function.

```python
num = int(input('Enter the number you want for multiplication table: '))
for i in range(1,11):
  print(i,'x',num,'=',i*num)
```
